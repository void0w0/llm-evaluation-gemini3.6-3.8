```python
import os
from google import genai

# APIキーを環境変数から読み込む
API_KEY = os.environ.get("GEMINI_API_KEY")
client = genai.Client(api_key=API_KEY)


# 質問する
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="自己紹介をしてください"
)
print(response.text)
```

```python
import pandas as pd
import time

# 評価するモデル
models = [
    "gemini-3.8-flash",
    "gemini-3.7-flash",
    "gemini-3.6-flash"
]

# 評価する質問
questions = [
    "トランスフォーマーアーキテクチャを料理に例えて説明してください"
]

results = []
for model_name in models:
    for question in questions:
        start = time.time()
        response = client.models.generate_content(
            model=model_name,
            contents=question
        )
        elapsed = round(time.time() - start, 2)
        answer = response.text.strip()

        # 回答を全文表示
        print(f"=== {model_name} ===")
        print(f"質問：{question}")
        print(f"応答時間：{elapsed}秒　文字数：{len(answer)}文字")
        print(answer)
        print()

        results.append({
            "モデル": model_name,
            "質問": question,
            "文字数": len(answer),
            "応答時間(秒)": elapsed
        })
```

## LLMモデル比較評価

### 概要
Google Geminiの3つのモデルに同じ質問を投げて回答を比較

### 使用モデル
- gemini-3.8-flash
- gemini-3.7-flash  
- gemini-3.6-flash

### 質問
- トランスフォーマーアーキテクチャを料理に例えて説明してください

### 結果の考察
- 3.6は網羅性に長けているが、使用する例の具材に一貫性は低い
- 3.7は専門性も保ちつつ、一番バランスがいい回答だった
- 3.8のモデルは、専門的知識をより抽象化に落とし込んだ時に直感的に理解しやすい
- 新しいモデルほど回答が短く簡潔になる傾向があり、古いモデルほど文字数が多いが応答時間は速かった。新しいモデルの解答時間が必ずしも短くなり、最適な回答をするとは限らないことがわかった。
